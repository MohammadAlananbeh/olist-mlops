```python
# outliers
plt.boxplot(training_set["total_price"].dropna())

plt.title("Total Price")
plt.ylabel("Value")

plt.show()
```

```python
# calculate IQR boundaries: using the IQR (Interquartile Range) method to identify potential outliers in the total_price
Q1 = training_set["total_price"].quantile(0.25)   # finds the 25th percentile. (25% of the observations fall below Q1 / 75% of observations are above Q1)
Q3 = training_set["total_price"].quantile(0.75)   # finds the 75th percentile. (75% of observations are below Q3 / 25% are above Q3)

IQR = Q3 - Q1  # Interquartile Range: It measures the spread of the middle 50% of your data. So the middle 50% of prices are between Q1 and Q3.

lower_bound = Q1 - 1.5 * IQR  # So anything below lower_bound would be considered a potential outlier.
upper_bound = Q3 + 1.5 * IQR  # So anything above upper_bound would be considered a potential outlier.

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

# count potential outliers:
outliers = training_set[
    (training_set["total_price"] < lower_bound) |
    (training_set["total_price"] > upper_bound)
]

len(outliers) # all rows where total_price is either below the lower bound OR above the upper bound 
              # (orders whose total_price falls outside the IQR boundaries. - outliers).

# training_set["total_price"] < lower_bound: means price is too low
# training_set["total_price"] > upper_bound: means price is too high

# Important: An outlier is not automatically an error.
# For example:
    # Very expensive order
    # could be a perfectly valid observation.
    # should investigate before removing anything.

```

```python
# reading olist_customers table
customers = pd.read_sql_query(
    "SELECT * FROM olist_customers",
    conn
)

customers.head()
```

```python
# merge the traingin_set with the customers table
training_set = training_set.merge(
    customers,
    on='customer_id',
    how='left'
)
# ---------------------------------------------
validation_set = validation_set.merge(
    customers,
    on='customer_id',
    how='left'
)
# ---------------------------------------------
testing_set = testing_set.merge(
    customers,
    on='customer_id',
    how='left'
)
```

```python
# reading olist_sellers table
sellers = pd.read_sql_query(
    "SELECT * FROM olist_sellers",
    conn
)

sellers.head()
```

```python
# merge the olist_order_items with the olist_sellers table
query = '''
SELECT
    oi.order_id,
    GROUP_CONCAT(DISTINCT s.seller_state) AS seller_state,
    GROUP_CONCAT(DISTINCT s.seller_zip_code_prefix) AS seller_zip_code_prefix,
    COUNT(DISTINCT oi.seller_id) AS seller_count
FROM olist_order_items oi
LEFT JOIN olist_sellers s
    ON oi.seller_id = s.seller_id
GROUP BY oi.order_id
'''

order_sellers = pd.read_sql_query(query, conn)

order_sellers.head()
```

```python
# # merge the traingin_set with the order_sellers table
training_set = training_set.merge(
    order_sellers,
    on='order_id',
    how='left'
)
# ----------------------------------------
validation_set = validation_set.merge(
    order_sellers,
    on='order_id',
    how='left'
)
# ----------------------------------------
testing_set = testing_set.merge(
    order_sellers,
    on='order_id',
    how='left'
)
```

```python
customer_state_analysis = (
    training_set
    .groupby("customer_state")
    .agg(
        orders=("order_id", "count"),
        late_rate=("is_late_label", "mean")
    )
    .sort_values("late_rate", ascending=False)
)

customer_state_analysis

# Do the same for seller state.
```

```python
seller_state_analysis = (
    training_set
    .groupby("seller_state")
    .agg(
        orders=("order_id", "count"), # Count the number of order_id values for each seller_state.
        late_rate=("is_late_label", "mean") # calculates the average of is_late_label for each seller_state.
    )
    .sort_values("late_rate", ascending=False)
)

seller_state_analysis
```

```python
# Customer–seller relationship
# A useful feature to investigate:
    # Same state?
    # Different state?

training_set["customer_seller_same_state"] = (
    training_set["customer_state"]
    == training_set["seller_state"]
).astype(int)
# -----------------------------------------------------
validation_set["customer_seller_same_state"] = (
    validation_set["customer_state"]
    == validation_set["seller_state"]
).astype(int)
# -----------------------------------------------------
testing_set["customer_seller_same_state"] = (
    testing_set["customer_state"]
    == testing_set["seller_state"]
).astype(int)
# -----------------------------------------------------

training_set.groupby("customer_seller_same_state")["is_late_label"].agg(
    ["count", "mean"]
)


# Potential finding:
# Orders where customer and seller are in different states
# have a higher late-delivery rate.
# If that pattern is strong, same_state could become a feature.

```

```python
# ZIP-code prefix and distance
# can explore:
    # customer_zip_code_prefix
    # seller_zip_code_prefix
# But :
    # ZIP-code prefix difference is not actual geographic distance.

training_set["customer_zip_code_prefix"] = pd.to_numeric(
    training_set["customer_zip_code_prefix"],
    errors="coerce"
)

training_set["seller_zip_code_prefix"] = pd.to_numeric(
    training_set["seller_zip_code_prefix"],
    errors="coerce"
)

training_set["zip_prefix_difference"] = (
    training_set["customer_zip_code_prefix"]
    - training_set["seller_zip_code_prefix"]
).abs()

# This may be useful for exploration, but it is only a rough proxy, not real distance.
# A better future approach could use geographical coordinates if available.

# /////////////////////////////////////////////////
validation_set["customer_zip_code_prefix"] = pd.to_numeric(
    validation_set["customer_zip_code_prefix"],
    errors="coerce"
)

validation_set["seller_zip_code_prefix"] = pd.to_numeric(
    validation_set["seller_zip_code_prefix"],
    errors="coerce"
)

validation_set["zip_prefix_difference"] = (
    validation_set["customer_zip_code_prefix"]
    - validation_set["seller_zip_code_prefix"]
).abs()

# /////////////////////////////////////////////////
testing_set["customer_zip_code_prefix"] = pd.to_numeric(
    testing_set["customer_zip_code_prefix"],
    errors="coerce"
)

testing_set["seller_zip_code_prefix"] = pd.to_numeric(
    testing_set["seller_zip_code_prefix"],
    errors="coerce"
)

testing_set["zip_prefix_difference"] = (
    testing_set["customer_zip_code_prefix"]
    - testing_set["seller_zip_code_prefix"]
).abs()
```
