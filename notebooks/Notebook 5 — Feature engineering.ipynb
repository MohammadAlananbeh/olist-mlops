```python
# For numerical columns, a good starting point is median imputation.
from sklearn.impute import (
    SimpleImputer,  # SimpleImputer is used to handle missing values dataset.
)

# can use SimpleImputer to replace missing values with:
    # Mean → average value
    # Median → middle value
    # Most frequent → most common value
    # Constant → a fixed value such as 0 or "Unknown"
# SimpleImputer = a tool that fills missing values before training machine learning model.

# Create median imputer , creates an imputer object called numeric_imputer.
    # SimpleImputer → Scikit-learn tool for handling missing values (NaN).
    # strategy="median" → Tells it to replace missing numerical values with the median of that column.
    # numeric_imputer → variable name for this imputer.
numeric_imputer = SimpleImputer(strategy="median")
# At this point, it has not calculated or replaced anything yet. still need to use .fit() and .transform().

# Fit on training data only - # Learn medians from TRAINING data only
numeric_imputer.fit(X_train[numeric_columns])  # This line tells the imputer to learn the median value of each numeric column using only X_train.
# .fit() → calculates and stores the replacement values.
# X_train[numeric_columns] → selects only the numerical columns from your training dataset.
# .fit() → It does not replace the missing values yet. It only calculates and stores the medians.
# so:
    #   fit() → learn the median from training data only
    #   transform() → use those same medians to fill missing values in train, validation, and test data 


# Transform train, validation, and test - # Apply the same learned medians to all splits
X_train[numeric_columns] = numeric_imputer.transform(
    X_train[numeric_columns]
)

X_val[numeric_columns] = numeric_imputer.transform(
    X_val[numeric_columns]
)

X_test[numeric_columns] = numeric_imputer.transform(
    X_test[numeric_columns]
)

# error: there is a missed column in val & test sets (solved) :
    # actual_delivery_days
    # delivery_difference_days
    # customer_zip_code_prefix
    # seller_zip_code_prefix
    # seller_count
    # customer_seller_same_state
    # zip_prefix_difference
```

```python
# For categorical columns:
# recommend "Unknown" because missing itself may contain useful information.
    # SimpleImputer(strategy="most_frequent")
    #  or
    # SimpleImputer(strategy="constant", fill_value="Unknown")

# from sklearn.impute import SimpleImputer

# Create categorical imputer: creates an imputer for categorical columns (If a categorical column has missing values (NaN), it replaces them with Unknown)
categorical_imputer = SimpleImputer(
    strategy="constant", # Replace every missing value with one fixed value that you specify.
    fill_value="Unknown"
)

# Fit on training data only
# fit() should be done only on X_train, then use the same fitted imputer to transform validation and test data to avoid data leakage.
categorical_imputer.fit(X_train[categorical_columns])

# Apply to all splits
X_train[categorical_columns] = categorical_imputer.transform(
    X_train[categorical_columns]
)

X_val[categorical_columns] = categorical_imputer.transform(
    X_val[categorical_columns]
)

X_test[categorical_columns] = categorical_imputer.transform(
    X_test[categorical_columns]
)
```

```python
from sklearn.preprocessing import OneHotEncoder

# OneHotEncoder: It converts categorical values into numerical columns that machine learning models can use.
# handle_unknown="ignore" is important: if a new category appears in the validation or test set that was not present in the training set, 
    # the encoder will not raise an error.

# How it works
# For every categorical column:
    # Finds the unique categories in the training data.
    # Creates one new column for each category.
    # Puts 1 in the matching category column.
    # Puts 0 in all other category columns.
# The encoder learns categories only from X_train, and then the same learned columns are used to transform X_val and X_test.

# Create the encoder
categorical_encoder = OneHotEncoder(
    handle_unknown="ignore",  # important because validation or test data may contain categories that were not present in training.
    sparse_output=False
)

# sparse_output=False: Python gives you a normal table-like array (Give me the complete table.)
# sparse_output=True (default): Python stores the same information in a memory-saving sparse format, because there are lots of zeros. 
    # (Don't waste memory storing all those zeros.)

# The encoder must be fitted only on the training set.

# Fit ONLY on training data
categorical_encoder.fit(X_train[categorical_columns]) # Teach the OneHotEncoder what categories exist in each categorical column, 
                                                        # using only the training data.
                                                        # It does not change X_train. It only learns and remembers the categories.
                                                        # fit() is "learn"

# Transform all three datasets
X_train_encoded = categorical_encoder.transform(
    X_train[categorical_columns]
)

X_val_encoded = categorical_encoder.transform(
    X_val[categorical_columns]
)

X_test_encoded = categorical_encoder.transform(
    X_test[categorical_columns]
)
```

Build the preprocessing pipeline
-------

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Numerical pipeline:
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# creates a two-step preprocessing pipeline.
    # Step 1 — Imputer: ("imputer", SimpleImputer(strategy="median"))
        # If a numerical feature has missing values: the imputer replaces NaN with the median of that column.
        # The median is calculated from training data when the pipeline is fitted.

    # Step 2 — Scaler: ("scaler", StandardScaler())
        # After missing values are handled, StandardScaler scales the numerical features.

```

```python
# Categorical pipeline:
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(
            strategy="constant",
            fill_value="Unknown"
        )),
        
        ("encoder", OneHotEncoder(
            handle_unknown="ignore"
        ))
    ]
)
```

```python
# Combine them:
# creates the main preprocessing object that combines numerical and categorical preprocessing.
preprocessor = ColumnTransformer( # combines columns
    transformers=[
        (
            "num",  # Take the columns listed in numerical_columns and send them through numeric_transformer.
            numeric_transformer,
            numeric_columns
        ),
        
        (
            "cat",  # Take the columns listed in categorical_columns and send them through categorical_transformer.
            categorical_transformer,
            categorical_columns
        )
    ]
)
```

```python
#                  X_train
#                     │
#           ┌─────────┴─────────┐
#           ↓                   ↓
#    Numerical columns     Categorical columns
#           │                   │
#           ↓                   ↓
# numeric_transformer   categorical_transformer
#           │                   │
#           ↓                   ↓
#  Impute + Scale       Impute + One-Hot Encode
#           │                   │
#           └─────────┬─────────┘
#                     ↓
#               Final features
```

```python
# This is one of the most important parts of notebook.
X_train_processed = preprocessor.fit_transform(X_train)

# Then only transform validation and test:
X_val_processed = preprocessor.transform(X_val)

X_test_processed = preprocessor.transform(X_test)
# ----------------------------------------------------
# Never do this:
    # preprocessor.fit_transform(X_val)
# or:
    # preprocessor.fit_transform(X_test)
# That would cause data leakage.
```
