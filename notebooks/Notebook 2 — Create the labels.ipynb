```python
# delete the orders which having no delivery date as its less than 3%
ml_orders = ml_orders[
    ml_orders["order_delivered_customer_date"].notna() &
    ml_orders["order_estimated_delivery_date"].notna()
].copy()  # .copy() → creates a clean independent DataFrame

ml_orders.shape
```

Create the delivery label  
label = 1  # Late:     when the actual delivery date is after the estimated date.  
label = 0  # not Late: when the actual delivery date is on time or before  

```python
#  creating the delivery label considering the missed delivery dates removed
# np.where(condition, value_if_true, value_if_false)
import numpy as np

ml_orders['is_late_label'] = np.where(
    ml_orders['order_delivered_customer_date'] >
    ml_orders['order_estimated_delivery_date'],
    1,
    0
)
```

```python
# Check real orders manually
# manual spot-check of 10 random orders to see whether 'is_late' label is correct.
# random_state=42: use 42 as the seed to determine which 10 rows (if need the same random values again, use the same random_state number)
check = ml_orders[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "is_late_label"
    ]
].sample(10, random_state=42)  # Randomly select 10 orders and the random selection reproducible (if run it again, will get the same 10 orders)

check
```

```python
# what is the count for late and not late orders
ml_orders["is_late_label"].value_counts()
```

```python
# what is the percent for late and not late orders
ml_orders["is_late_label"].value_counts(normalize=True) * 100
```

```python
# Look at class distribution
class_distribution = (
    ml_orders["is_late_label"]
    .value_counts()
    .rename(index={0: "On time", 1: "Late"})
    .to_frame("count") 
    # Convert the Series to a DataFrame:
    # value_counts() normally returns a Series. 
    # so, .to_frame("count") converts it into a DataFrame and names the column "count".
)

class_distribution["percentage"] = (
    class_distribution["count"]
    / class_distribution["count"].sum()
    * 100
)

class_distribution
```

```python
# Decide whether there is class imbalance (Class imbalance means that target classes have very different numbers of observations.)
# check whether one class is much larger than the other
counts = ml_orders["is_late_label"].value_counts()

majority = counts.max()
minority = counts.min()

imbalance_ratio = majority / minority

print(f"Majority class: {majority:,}")
print(f"Minority class: {minority:,}")
print(f"Imbalance ratio: {imbalance_ratio:.2f}:1")
```
